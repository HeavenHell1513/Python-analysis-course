Класс

```python
# Создание класса
class Chocolate:
  def __init__(self, cost, taste = 'None'):
    self.cost = float(cost)
    self.taste = str(taste)
    # Проверка цены
    if self.cost <= 0:
      raise ValueError('Меньше или равно 0')
  # Функция по описанию вкуса шоколада
  def show_my_chocolate(self):
    if (self.taste) == 'None' :
      print ('Обычный шоколадный батончик')
    else :
      print ('Шоколадный батончик, добавка:', self.taste)
# Отображение элементов класса в читабельном формате
  def __str__(self):
    return f'Шоколадный батончик, {self.cost} рублей'
```

Данные для проверки

```python
mars = Chocolate(3,'Kofe')
print(mars)
mars.show_my_chocolate()

snikers = Chocolate(5)
snikers.show_my_chocolate()
```

Данные с малой ценой

```python
bounty = Chocolate(-1,'Sahar')
print(bounty)
bounty.show_my_chocolate()
```
